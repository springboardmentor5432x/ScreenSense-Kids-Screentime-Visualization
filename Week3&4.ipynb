```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("preprocessed_kids_screentime.csv")   


```

```python
#week-3
#1.Distribution of dailt screen time-histogram
plt.figure()
plt.hist(df['Avg_Daily_Screen_Time_hr'], bins=10)
plt.xlabel("Daily Screen Time (hours)")
plt.ylabel("Number of Students")
plt.title("Distribution of Daily Screen Time")
plt.show()
```

```python
#2.Distribution of age bands-barchart
plt.figure()
df['Age_Band'].value_counts().plot(kind="bar")
plt.xlabel("Age Band")
plt.ylabel("Count")
plt.title("Age Band Distribution")
plt.show()
```

```python
#3.most used devices
plt.figure()
df['Primary_Device'].value_counts().plot(kind="bar")
plt.xlabel("Device Type")
plt.ylabel("Number of Students")
plt.title("Most Used Device Types")
plt.show()
```

```python
#4.comparing rural vs urban
plt.figure()
df['Urban_or_Rural'].value_counts().plot(kind="bar")
plt.xlabel("Region")
plt.ylabel("Number of Students")
plt.title("Urban vs Rural User Distribution")
plt.show()
```

```python
#Univariate Analysis
#5.comparing screentime by gender-boxplot
plt.figure()
df.boxplot(column='Avg_Daily_Screen_Time_hr', by='Gender')
plt.title("Screen Time by Gender")
plt.suptitle("")
plt.xlabel("Gender")
plt.ylabel("Daily Screen Time (hours)")
plt.show()
```

```python
#6.screentime by age band
plt.figure()
df.boxplot(column='Avg_Daily_Screen_Time_hr', by='Age_Band')
plt.title("Screen Time by Age Band")
plt.suptitle("")
plt.xlabel("Age Band")
plt.ylabel("Daily Screen Time (hours)")
plt.show()
```

```python
#7.Primary device used across age bands-grouped barchart
device_age = pd.crosstab(df['Age_Band'], df['Primary_Device'])
device_age.plot(kind='bar')
plt.xlabel("Age Band")
plt.ylabel("Count")
plt.title("Device Usage Across Age Groups")
plt.show()
```

```python
#week4
#8.Exceeded time limit
plt.figure()
df.groupby('Exceeded_Recommended_Limit')['Avg_Daily_Screen_Time_hr'].mean().plot(kind='bar')
plt.xlabel("Exceeded Limit (True/False)")
plt.ylabel("Average Screen Time (hours)")
plt.title("Screen Time based on Recommended Limit (Bar Chart)")
plt.show()
```

```python
#11
age_numeric = df.groupby('Age')['Avg_Daily_Screen_Time_hr'].mean()

plt.figure()
plt.plot(age_numeric.index, age_numeric.values, marker='o')
plt.xlabel("Age")
plt.ylabel("Average Screen Time (hours)")
plt.title("Screen Time Trend by Age (Line Plot)")
plt.show()
```
